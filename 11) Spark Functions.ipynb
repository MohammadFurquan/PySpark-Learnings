# 1) withColumn()

Used to create a new column or modify an existing column in a DataFrame.

### Syntax

```python
df.withColumn(column_name, expression)
```

### Parameters

- `column_name` → Name of the column
- `expression` → Logic/expression to be applied

### Example

```python
df.withColumn("discount", F.col("amount") * 0.1)
```

# 4) regexp_replace()

Replaces matching text using a regular expression.

### Syntax

```python
F.regexp_replace(column, pattern, replacement)
```

### Parameters

- `column` → Target column
- `pattern` → Regex pattern to search
- `replacement` → Replacement value

### Example

```python
F.regexp_replace(F.col("brand_code"), r'[^A-Za-z0-9]', '')
```

---

# 5) replace()

Replaces existing values in a DataFrame.

### Syntax

```python
df.replace(to_replace, value=None, subset=None)
```

### Parameters

- `to_replace` → Value or mapping to replace
- `value` → Replacement value
- `subset` → Target columns

### Example

```python
df.replace(
    {"NA": "UNKNOWN"},
    subset=["category_code"]
)
```
