# ACID Properties in Databricks (Delta Lake)

ACID properties ensure that data remains **accurate, consistent, and reliable** even when multiple users or jobs are reading and writing data simultaneously.

Databricks provides ACID transactions through **Delta Lake**.

---

# What Does ACID Stand For?

```text
A → Atomicity
C → Consistency
I → Isolation
D → Durability
```

---

# 1. Atomicity

Atomicity means:

> Either the entire transaction succeeds or the entire transaction fails.

There is no partial update.

### Example

Suppose a banking transaction transfers ₹1000:

```text
Account A → -1000
Account B → +1000
```

Possible outcomes:

```text
✅ Both operations succeed
```

or

```text
✅ Both operations are rolled back
```

Not:

```text
❌ Account A debited
❌ Account B not credited
```

### Benefit

✅ Prevents partial writes

✅ Prevents data corruption

---

# 2. Consistency

Consistency means:

> Data always remains in a valid state before and after a transaction.

### Example

Table Rule:

```text
salary > 0
```

Invalid Data:

```text
salary = -5000
```

Such invalid records should not violate business rules and constraints.

### Benefit

✅ Maintains data integrity

✅ Ensures valid data

---

# 3. Isolation

Isolation means:

> Multiple transactions can run simultaneously without interfering with each other.

### Example

Two users update the same table:

```text
User A → Updating Data
User B → Reading Data
```

User B should not see incomplete or partially written data.

### Benefit

✅ Prevents dirty reads

✅ Prevents inconsistent results

✅ Supports concurrent users

---

# 4. Durability

Durability means:

> Once a transaction is committed, it is permanently stored.

Even if:

```text
Cluster crashes
System restarts
Power failure occurs
```

the committed data remains safe.

### Example

```sql
INSERT INTO sales VALUES (101, 1000);
```

Once committed:

```text
✅ Data is permanently saved
```

### Benefit

✅ Protects against data loss

✅ Reliable storage

---

# What is Time Travel in Databricks?

**Time Travel** is a Delta Lake feature that allows you to view or restore a table from a previous version.

Think of it as:

```text
Going back in time
to see how the table looked earlier.
```

---

# Why Do We Need Time Travel?

Suppose your table contains:

```text
id | name
----------
1  | Arif
2  | Rahul
3  | Amit
```

Someone accidentally runs:

```sql
DELETE FROM customers;
```

Now all records are deleted.

Normally:

```text
Data Lost ❌
```

But with Delta Lake Time Travel:

```text
Data Can Be Recovered ✅
```

because Delta keeps previous versions of the table.

---

# How Time Travel Works

Every operation creates a new version.

### Initial Table

```text
Version 0
```

| id | name |
|----|------|
| 1 | Arif |
| 2 | Rahul |

---

### Insert New Record

```sql
INSERT INTO customers VALUES (3, 'Amit');
```

Creates:

```text
Version 1
```

| id | name |
|----|------|
| 1 | Arif |
| 2 | Rahul |
| 3 | Amit |

---

### Delete Record

```sql
DELETE FROM customers WHERE id = 2;
```

Creates:

```text
Version 2
```

| id | name |
|----|------|
| 1 | Arif |
| 3 | Amit |

---

Delta keeps:

```text
Version 0
Version 1
Version 2
```

---

# View Table History

```sql
DESCRIBE HISTORY customers;
```

Output:

```text
Version
0
1
2
```

along with:

```text
Operation
User
Timestamp
```

---

# Read Older Version

Suppose current version is:

```text
Version 2
```

and you want to see:

```text
Version 1
```

```sql
SELECT *
FROM customers VERSION AS OF 1;
```

Result:

```text
1 | Arif
2 | Rahul
3 | Amit
```

---

# Read Data by Timestamp

Instead of version number:

```sql
SELECT *
FROM customers
TIMESTAMP AS OF '2025-01-01 10:00:00';
```

Delta returns the table exactly as it existed at that time.

---

# Real-Life Example

Suppose a Sales table contains:


A developer accidentally executes:


Current version:

```text

```

Before deletion:

You can recover data using:

```sql
SELECT *
```

# Interview Answer

> Time Travel is a Delta Lake feature that allows users to access and query previous versions of a table using a version number or timestamp. Every write operation creates a new version of the table, and Delta Lake maintains the transaction history. Time Travel is commonly used for data recovery, auditing, debugging, and analyzing historical data.


````python
# Time travel is like a git, it maintains history after every write operation, and we can revert the commit like this here also we can do the same.


# Query data as it existed at versison 5
Select * from my_table VERSION AS OF 5

# Query data as it existed at a specific timestamp
Select * from my_table TIMESTAMP AS OF ‘2024-12-01 10:00:00’

# Restore table to previous version
RESTORE table my_table TO VERSION AS OF 10

````

# What is Delta Lake?

**Delta Lake** is an open-source storage layer that brings Data Warehouse features to a Data Lake.

It provides advanced capabilities such as:

✅ ACID Transactions

✅ Time Travel

✅ Schema Enforcement

✅ Audit History

✅ MERGE, UPDATE, and DELETE support

✅ Reliable and Consistent Data Processing

---

# Why Do We Need Delta Lake?

Without Delta Lake:

```text
S3 / ADLS / GCS
      ↓
 Parquet Files
```

You only have files stored in the Data Lake.

Problems:

❌ No ACID transactions

❌ No Time Travel

❌ Difficult to handle concurrent writes

❌ Risk of data corruption

❌ Limited support for updates and deletes

---

# How Delta Lake Solves These Problems

Delta Lake adds a transaction log called:

```text
_delta_log
```

This log tracks:

```text
Who modified the data
When it was modified
What was inserted
What was updated
What was deleted
```

This makes data processing reliable and consistent.

---

# Key Features of Delta Lake

## 1. ACID Transactions

Delta Lake guarantees:

```text
Atomicity
Consistency
Isolation
Durability
```

### Example

```sql
UPDATE sales
SET amount = 1000
WHERE id = 1;
```

Possible outcomes:

```text
✅ Complete Update
```

or

```text
✅ Complete Rollback
```

Never:

```text
❌ Partial Update
```

---

## 2. Time Travel

Delta Lake stores table history and versions.

Example:

```sql
SELECT *
FROM sales VERSION AS OF 5;
```

This allows you to view how the table looked in Version 5.

### Benefits

✅ Recover accidentally deleted data

✅ Audit changes

✅ Compare historical data

✅ Reproduce old reports

---

## 3. Schema Enforcement

Expected Schema:

```text
salary → Double
```

Incoming Data:

```text
salary = "abc"
```

Delta Lake validates the schema and rejects invalid data.

### Benefits

✅ Better data quality

✅ Prevents bad data from being written

---

## 4. UPDATE, DELETE, and MERGE Support

Traditional Parquet tables do not efficiently support:

```sql
UPDATE
DELETE
MERGE
```

Delta Lake supports all of them.

### Examples

```sql
UPDATE sales
SET amount = 2000
WHERE id = 101;
```

```sql
DELETE FROM sales
WHERE id = 101;
```

```sql
MERGE INTO customers
USING updates
ON customers.id = updates.id
```

---

## 5. Data Versioning

Every write operation creates a new table version.

Example:

```text
Version 0
Version 1
Version 2
Version 3
```

Because all versions are tracked, Delta Lake can support Time Travel.


### Important


is the heart of Delta Lake.
