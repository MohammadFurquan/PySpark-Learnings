# Data Warehouse vs Data Lake

Both are used to store data, but they serve different purposes.

---

# What is a Data Warehouse?

A **Data Warehouse** is a centralized repository designed for storing **structured and processed data** for reporting, analytics, and business intelligence.

Think of it as:

```text
Clean Data
     ↓
Reporting
     ↓
Business Decisions
```

### Examples

- Snowflake
- Amazon Redshift
- Azure Synapse
- Google BigQuery
- Teradata

---

## Data Warehouse Characteristics

✅ Structured data only

✅ Schema enforced before loading data

✅ Fast SQL queries

✅ Optimized for reporting

✅ Supports ACID transactions

❌ Expensive storage

❌ Not ideal for unstructured data

---

## Example

Suppose an e-commerce company wants to analyze:

```text
Daily Sales
Monthly Revenue
Top Products
Customer Trends
```

Data is first cleaned and transformed:

```text
Raw Data
    ↓
ETL Process
    ↓
Data Warehouse
    ↓
Power BI / Tableau
```

Example Query:

```sql
SELECT region, SUM(sales)
FROM sales
GROUP BY region;
```

---

## When to Use a Data Warehouse?

Use a Data Warehouse when:

✅ Data is already cleaned

✅ Mostly structured data

✅ Business reporting is the main requirement

✅ Fast SQL analytics is required

✅ Dashboards and KPIs need to be generated

---

# What is a Data Lake?

A **Data Lake** is a storage repository that can store **structured, semi-structured, and unstructured data** in its raw format.

Think of it as:

```text
Store Everything First
Analyze Later
```

### Examples

- Amazon S3
- Azure Data Lake Storage (ADLS)
- Google Cloud Storage (GCS)

---

## Data Lake Characteristics

✅ Stores any type of data

✅ Cheap storage

✅ Massive scalability

✅ Ideal for big data and ML

✅ Schema applied during reading

❌ Slower analytics compared to warehouses

❌ Governance can be challenging

---

## Example

An e-commerce company collects:

```text
Sales Data
Customer Data
Application Logs
JSON Files
Images
Videos
Clickstream Data
```

All data is stored directly in the Data Lake.

```text
Raw Data
     ↓
Data Lake (S3 / ADLS)
```

No need to transform data before storing it.

---

## When to Use a Data Lake?

Use a Data Lake when:

✅ You need to store raw data

✅ Data comes from multiple sources

✅ You have unstructured data

✅ Building ML/AI solutions

✅ Long-term data storage is needed

✅ Cost-effective storage is important

# Schema-on-Write vs Schema-on-Read

The main difference is **when the schema is applied and validated**.

---

# Schema-on-Write

Schema-on-Write means:

> Data is validated against a predefined schema **before it is stored**.

### Flow

```text
Data
 ↓
Schema Validation
 ↓
Store Data
```

### Example

Expected Schema:

```text
id      → Integer
name    → String
salary  → Double
```

Incoming Data:

```text
id = 101
name = Arif
salary = "abc"
```

Since `salary` should be a Double but a String is received, the data will be rejected.

### Advantages

✅ High data quality

✅ Consistent data

✅ Reliable reporting

✅ Faster analytics

### Used In

```text
Data Warehouses
```

Examples:

- Snowflake
- Amazon Redshift
- Azure Synapse
- SQL Server

---

# Schema-on-Read

Schema-on-Read means:

> Data is stored first, and the schema is applied only when the data is read.

### Flow

```text
Data
 ↓
Store Data
 ↓
Read Data
 ↓
Apply Schema
```

### Example

Incoming Data:

```json
{
  "id": 101,
  "name": "Arif",
  "salary": "abc"
}
```

The data is stored successfully in the Data Lake.

Later, while reading:

```python
spark.read.schema(my_schema).json("/data")
```

Spark attempts to apply the schema at read time.

### Advantages

✅ Very flexible

✅ Can store any type of data

✅ Suitable for Big Data workloads

✅ Great for Data Science and Machine Learning

### Used In

```text
Data Lakes
```

Examples:

- Amazon S3
- Azure Data Lake Storage (ADLS)
- Google Cloud Storage (GCS)

# What is Medallion Architecture?

**Medallion Architecture** is a data design pattern used in Databricks and Data Engineering to organize data into multiple layers that gradually improve data quality.

The data typically moves through:

```text
Bronze Layer
     ↓
Silver Layer
     ↓
Gold Layer
```

As data moves from Bronze → Silver → Gold:

✅ Data quality improves

✅ Data becomes more business-ready

✅ Easier governance and maintenance

✅ Better performance

---

# Why Do We Need Medallion Architecture?

Suppose data is coming from:

```text
CSV Files
JSON Files
APIs
Databases
Streaming Sources
```

Raw data may contain:

```text
Duplicates
Null Values
Incorrect Records
Bad Formats
Missing Data
```

Instead of directly exposing raw data to business users, we process it through layers.

---

# Bronze Layer (Raw Data)

The Bronze layer stores data exactly as received from the source.

### Characteristics

✅ Raw data

✅ No transformations

✅ Historical data retained

✅ Used for auditing and reprocessing

### Example

Source Data:

```text
id,name,salary
1,Arif,50000
2,Rahul,NULL
3,Amit,ABC
```

Stored exactly as received.

```text
Raw Data
     ↓
Bronze Table
```

### Purpose

> Never lose the original source data.

---

# Silver Layer (Cleaned Data)

The Silver layer contains cleaned and validated data.

### Operations Performed

✅ Remove duplicates

✅ Handle null values

✅ Data validation

✅ Standardize formats

✅ Join datasets

✅ Apply business rules

### Example

Bronze:

```text
1,Arif,50000
2,Rahul,NULL
3,Amit,ABC
```

Silver:

```text
1,Arif,50000
```

Invalid records removed or corrected.

### Purpose

> Create trusted and high-quality datasets.

---

# Gold Layer (Business-Ready Data)

The Gold layer contains aggregated and business-friendly data.

### Operations Performed

✅ Aggregations

✅ KPIs

✅ Business metrics

✅ Reporting tables

✅ Dashboard tables

### Example

Silver Data:

```text
Customer Transactions
```

Gold Data:

```text
Region     Total Sales
North      1,000,000
South        750,000
East         500,000
```

### Purpose

> Provide data for reports, dashboards, and business users.

---

# Interview Answer

> Medallion Architecture is a data design pattern used in Databricks and Lakehouse environments to progressively improve data quality through multiple layers. It consists of Bronze, Silver, and Gold layers. The Bronze layer stores raw source data, the Silver layer contains cleaned and validated data, and the Gold layer contains aggregated, business-ready datasets used for reporting and analytics. Medallion Architecture improves data quality, governance, reusability, scalability, and simplifies debugging and maintenance of data pipelines.

Operations:

- Remove duplicates
- Handle NULL values
- Fix data types
- Standardize formats
- Apply business rules


### Purpose

✅ Trusted data

✅ Clean data

✅ Used by data engineers and analysts

> Silver = "Clean and validated data."

---

# Gold Layer (Business Ready Data)

Create business metrics and aggregations.

### Purpose

✅ Dashboards

✅ Executive summaries

Think:

> Gold = "Data ready for business users."

# 51. Complete End-to-End Interview Answer

### Answer

In a Databricks Lakehouse architecture, raw data is first ingested into the Bronze layer, where it is stored without modification. The data is then cleansed, validated, deduplicated, and standardized in the Silver layer. Finally, business metrics, KPIs, and aggregated datasets are created in the Gold layer for reporting and analytics. This Medallion Architecture improves data quality, governance, scalability, and maintainability while leveraging the capabilities of a modern Lakehouse platform.
