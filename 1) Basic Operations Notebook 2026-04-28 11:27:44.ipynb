```python
from pyspark.sql.functions import col

# list the movies whose rating is 0 or 9.2 then accept 
display(df.filter((col("imdb_rating") == 9) | (col("imdb_rating") == 9.2)).select("title", "imdb_rating"))
```

### Row Filtering

```python
# # this is first way
# df_filtered = (df.filter((df.release_year >= 2000) & (df.release_year <= 2010)))
df_filtered = (df.filter((df["release_year"] >= 2000) & (df.release_year <= 2010)))
display(df_filtered)
# Note:  df.release_year and df["release_year"] are the same
```

```python
# this is second way
from pyspark.sql.functions import col

# filter rows where release_year is between 2000 and 2010(inclusive)
df_filt = df.filter((col("release_year") >= 2000) & (col("release_year") <= 2010))
# show a few rows
df_filt.show()
```

```python
# this is third way
df_filtered = df.filter(col("release_year").between(2000,2010))
display(df_filtered)
```

```python
# All marvel movies
df_marvel = df.filter(col("studio") == "Marvel Studios")
df_marvel.show()
```

```python
# how many distinct movies are there
unique_industries = df.select("industry").distinct()
display(unique_industries)
```

```python
unique_lang = df.select("language").distinct()
display(unique_lang)
```

```python
# Adding 1 column of profit that is revenue - budget and as we want to add it in same df so assigning it into same column
df = df.withColumn('profit', col('revenue') - col('budget'))
# withColumn : Iska use naya column add karne ya existing column ko modify/update karne ke liye hota hai.
display(df)
```

```python
df = df.withColumnRenamed("revenue", "total_revenue")

df.printSchema()
```
